```python
import pandas as pd
import warnings
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.feature_selection import SelectFromModel

warnings.filterwarnings("ignore")

# Read data
data = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/T Cell Antigen Datasets/Features/Features Extraxtion/DS-2/CrossAttention_fused_features.csv")

# Features and label
X = data.iloc[:, :-1]
y = data.iloc[:, -1]

print("Original data shape:", data.shape)
print("Feature shape:", X.shape)
print("Label shape:", y.shape)

# Train GBDT model
clf = GradientBoostingClassifier(random_state=10)
clf.fit(X, y)

# Select important features
selector = SelectFromModel(clf, prefit=True, threshold="mean")
support = selector.get_support()
selected_cols = X.columns[support]

# Create selected feature dataframe
X_selected = X[selected_cols]

# Add label column back
selected_data = pd.concat([X_selected, y], axis=1)

# Save selected features to CSV
output_path = "GBDT_Selected_Features-TS.csv"
selected_data.to_csv(output_path, index=False)

print("Number of selected features:", len(selected_cols))
print("Selected feature names:")
print(list(selected_cols))
print("Selected feature data shape:", selected_data.shape)
print("Saved file at:", output_path)
```
